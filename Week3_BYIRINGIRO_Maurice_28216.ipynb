```python
# Exercise 1.1 – Grading ladder
def print_grade(score):
    if score >= 85:
        print("A")
    elif score >= 70:
        print("B")
    elif score >= 60:
        print("C")
    elif score >= 50:
        print("D")
    else:
        print("F")

for s in [91, 76, 64, 52, 43]:
    print(f"Score {s} → ", end="")
    print_grade(s)
```

```python
# Exercise 1.2 – Combined conditions
def check_validation(score, attendance):
    if score >= 50 and attendance >= 75:
        print("Validated")
    elif score >= 50:
        print("Blocked: low attendance")
    else:
        print("Failed")

check_validation(76, 92)
check_validation(76, 68)
check_validation(43, 95)
```

```python
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

for s in scores:
    if s >= 80:
        print(f"Top score: {s}")
```

```python

total = 0
count = 0
passed = 0
failed = 0

for s in scores:
    total += s
    count += 1
    if s >= 50:
        passed += 1
    else:
        failed += 1

average = total / count

print("Total:", total)
print("Count:", count)
print("Average:", average)
print("Passed:", passed)
print("Failed:", failed)
```

```python
# Exercise 2.3 – while loop
savings = 0
months = 0
target = 250000
monthly = 15000

while savings < target:
    savings += monthly
    months += 1

print(f"You need {months} months to reach {target} RWF")
```

```python
# Exercise 3.1
def get_grade(score):
    """Return the letter grade for a numeric score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

print(get_grade(91), get_grade(76), get_grade(43))
```

```python
# Exercise 3.2
def pass_rate(score_list):
    """Return the percentage of scores ≥ 50 using an accumulator."""
    total = 0
    passed = 0
    for s in score_list:
        total += 1
        if s >= 50:
            passed += 1
    return (passed / total) * 100


print(pass_rate(scores))  
```

```python
# Exercise 4.1
with open("week3_students.csv") as f:
    lines = f.readlines()

print("Number of lines:", len(lines))          
print("Header:", lines[0].strip())
print("First student:", lines[1].strip())
# Why 41 lines?
```

```python
# Exercise 4.2
names = []
for line in lines[1:]:
    parts = line.strip().split(",")
    names.append(parts[1])

print("Count of names:", len(names))
print("First 5 names:", names[:5])
```

```python
# Exercise 5.1 – Survive the mess (find the 3 broken records)
total = 0
valid = 0
bad = 0
bad_ids = []

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        total += score
        valid += 1
    except ValueError:
        bad += 1
        bad_ids.append(parts[0])

print("Valid records:", valid)
print("Bad records:", bad)
print("Broken student IDs:", bad_ids)
print("Average of valid scores:", round(total / valid, 1))
```

```python
# Exercise 5.2 – Grades & top student
passed = 0
failed = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        name = parts[1]
        if score >= 50:
            passed += 1
        else:
            failed += 1
        if score > best_score:
            best_score = score
            best_name = name
    except ValueError:
        continue

print("Passed:", passed)
print("Failed:", failed)
print("Top student:", best_name, "with score", best_score)
print("Letter grade:", get_grade(best_score))
```

```python
# Exercise 5.3 – Average by district
sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        district = parts[5]
        sums[district] = sums.get(district, 0) + score
        counts[district] = counts.get(district, 0) + 1
    except ValueError:
        continue

print("Average score by district:")
for d in sorted(sums):
    avg = sums[d] / counts[d]
    print(f"  {d}: {avg:.1f}")
```

```python
# Bonus – Write report file
with open("week3_report.txt", "w") as f:
    f.write(f"Student count: 40\n")
    f.write(f"Valid: {valid}, Bad: {bad}\n")
    f.write(f"Average (valid): {round(total/valid, 1)}\n")
    f.write(f"Top student: {best_name} ({best_score})\n")
    f.write(f"Broken records: {', '.join(bad_ids)}\n")

with open("week3_report.txt") as f:
    print(f.read())
```
